```python
# Revisão documental — 20/10/2026.
# Saídas históricas preservadas. As saídas de C4 e C6 não correspondem às instruções atuais: os labels
# são impressos em C5 e os centros em C8. Execução integral desta revisão no Colab pendente.

# Importa a classe KMeans do módulo cluster do scikit-learn
from sklearn.cluster import KMeans
# Importa a biblioteca numpy para operações numéricas, com o alias np
import numpy as np
```

```python
# Cria um array numpy com os dados de entrada para o algoritmo de clustering
X = np.array([[1, 2], [1, 4], [1, 0],[4, 2], [4, 4], [4, 0]])
```

```python
# Cria uma instância do modelo KMeans com 2 clusters e estado aleatório fixo para reprodutibilidade
kmeans = KMeans(n_clusters=2, random_state=0, n_init=10).fit(X)
```

```python
# Imprime os rótulos de cluster atribuídos a cada ponto no conjunto de dados X
print(kmeans.labels_)
```

```python
# Prevê o cluster para novos pontos de dados [[0, 0], [4, 4]]
kmeans.predict([[0, 0], [4, 4]])
```

```python
# Imprime os rótulos de cluster atribuídos aos novos pontos
print(kmeans.predict([[0, 0], [4, 4]]))
```

```python
# Imprime as coordenadas dos centros dos clusters encontrados pelo algoritmo
print(kmeans.cluster_centers_)
```
