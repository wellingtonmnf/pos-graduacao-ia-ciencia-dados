```python
# Importa a classe CountVectorizer do módulo feature_extraction.text da biblioteca scikit-learn
from sklearn.feature_extraction.text import CountVectorizer
```

```python
# Define um corpus (conjunto de documentos) como uma lista de strings
corpus = [
     'This is the first document.',
     'This document is the second document.',
     'And this is the third one.',
     'Is this the first document?',
]
```

```python
# Cria uma instância do CountVectorizer
vectorizer = CountVectorizer()
# Ajusta o vectorizer ao corpus e transforma o corpus em uma matriz de contagem de termos
X = vectorizer.fit_transform(corpus)
```

```python
# Imprime os nomes das features (palavras únicas) encontradas no corpus
print(vectorizer.get_feature_names_out())
```

```python
# Imprime a matriz de contagem de termos como um array denso
print(X.toarray())
```
