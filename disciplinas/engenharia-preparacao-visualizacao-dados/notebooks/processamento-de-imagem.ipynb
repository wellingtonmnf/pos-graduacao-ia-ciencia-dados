## Leitura da Imagem

```python
# Leitura de uma imagem
# Importar bibliotecas
import cv2
import numpy as np
from IPython.display import Image
import matplotlib.pyplot as plt
import matplotlib
```

```python
# Leitura da imagem
# img = cv2.imread(str(IMAGENS_DIR / 'imagem.png'))
# img = cv2.imread(str(IMAGENS_DIR / 'alguns.jpg'))
img = cv2.imread(str(IMAGENS_DIR / 'muitos.jpg'))
```

```python
# Apresentação da imagem na tela
plt.figure(figsize=(10, 10), dpi=100)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.show()
```

## Detecção de faces

```python
# Detecção de faces
# O tamanho de cada figura é definido na visualização, em polegadas.
```

```python
# Classificador construído para detectar faces
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

if face_cascade.empty():
    print("Error loading cascade classifier")
else:
    print("Cascade classifier loaded successfully")
```

```python
# Transformando a imagem em escala de cinza
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
# Processo para detectar as faces na imagem
# Busca e detecção
faces = face_cascade.detectMultiScale(
 gray,
 scaleFactor=1.1,
 minNeighbors=5,
 minSize=(30, 30)
)
```

```python
# Processamento das faces
cont = 0
for (x,y,w,h) in faces:
 img = cv2.rectangle(img,(x,y),(x+w,y+h),(255,0,0),2)
 roi_gray = gray[y:y+h, x:x+w]
 roi_color = img[y:y+h, x:x+w]
 cont=cont + 1
```

```python
# Salvamento e visualização
#cv2.imwrite(str(RESULTADOS_DIR / 'aragorn.png'),img)
#cv2.imwrite(str(RESULTADOS_DIR / 'poucos.jpg'),img)
cv2.imwrite(str(RESULTADOS_DIR / 'todos.jpg'),img)
plt.figure(figsize=(10, 10), dpi=100)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.show()
```

## Carregamento e Análise

```python
# Carrega a imagem com seus dados de altura, largura e canais
#img = cv2.imread(str(RESULTADOS_DIR / 'aragorn.png'),-1)
#img = cv2.imread(str(RESULTADOS_DIR / 'poucos.jpg'),-1)
img = cv2.imread(str(RESULTADOS_DIR / 'todos.jpg'),-1)
height, width, channels = img.shape
# Cria a tupla colors (BGR) e cria a lista features
colors = ('b', 'g', 'r')
features=[]
```

```python
# Cria a máscara
mask = np.zeros(img.shape[:2], np.uint8)
mask[int(height*0.1):int(height*0.9), 0:int(width*0.6)] = 255
masked_img = cv2.bitwise_and(img,img,mask = mask)
```

```python
# Exemplo Doc OpenCV
# Carrega a imagem com a adição da máscara
plt.figure(figsize=(8, 4), dpi=100)
for i, col in enumerate(colors):
 hist_mask = cv2.calcHist([img], [i], mask, [256], [0, 256])
 plt.plot(hist_mask, color=col)
 plt.xlim([0, 256])
plt.show()
```
