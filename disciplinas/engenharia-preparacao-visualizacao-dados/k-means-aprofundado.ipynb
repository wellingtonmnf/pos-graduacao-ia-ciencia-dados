```python
from sklearn.cluster import KMeans
import numpy as np
```

```python
# Dados originais
X = np.array([[1, 2], [1, 4], [1, 0], [4, 2], [4, 4], [4, 0]])

print("=== DADOS ORIGINAIS ===")
print("X =", X)
```

```python
# Executar o K-Means exatamente como no exemplo
kmeans = KMeans(n_clusters=2, random_state=0, n_init=3).fit(X)

print("=== RESULTADOS FINAIS ===")
print("Labels (grupos):", kmeans.labels_)
print("Centroides finais:")
print(kmeans.cluster_centers_)
print("Inércia final:", kmeans.inertia_)
```

```python
# Predição para novos pontos
predicoes = kmeans.predict([[0, 0], [4, 4]])
print("Predições para [[0, 0], [4, 4]]:", predicoes)
```

```python
# Vamos investigar o processo de inicialização usando métodos disponíveis publicamente
print("=== INVESTIGANDO A INICIALIZAÇÃO ===")

# Vamos também testar o que acontece com diferentes métodos de inicialização
print("=== TESTANDO DIFERENTES INICIALIZAÇÕES ===")
```

```python
# Method 1: k-means++ (default)
kmeans_kpp = KMeans(n_clusters=2, random_state=0, init='k-means++').fit(X)
print("Com k-means++ init:")
print("Labels:", kmeans_kpp.labels_)
print("Centroides:", kmeans_kpp.cluster_centers_)
print("Inércia:", kmeans_kpp.inertia_)
```

```python
# Method 2: random
kmeans_random = KMeans(n_clusters=2, random_state=0, init='random').fit(X)
print("Com random init:")
print("Labels:", kmeans_random.labels_)
print("Centroides:", kmeans_random.cluster_centers_)
print("Inércia:", kmeans_random.inertia_)
```

```python
# Vamos ver como o random_state afeta a sequência de números
print("=== SEQUÊNCIA DE NÚMEROS PSEUDOALEATÓRIOS ===")
np.random.seed(0)
print("Primeiros 10 números com seed=0:")
print(np.random.random(10))

np.random.seed(0)
print("Mesma sequência repetida:")
print(np.random.random(10))
```

```python
print("=== DISTÂNCIAS DOS PONTOS AOS CENTROIDES FINAIS ===")
centroides = kmeans.cluster_centers_
for i, ponto in enumerate(X):
    dist_0 = np.linalg.norm(ponto - centroides[0])
    dist_1 = np.linalg.norm(ponto - centroides[1])
    grupo = kmeans.labels_[i]
    print(f"Ponto {ponto}: dist_centro_0={dist_0:.3f}, dist_centro_1={dist_1:.3f}, grupo={grupo}")
```
