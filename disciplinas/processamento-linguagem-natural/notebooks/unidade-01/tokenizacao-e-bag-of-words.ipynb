```python
import nltk
# Faz o download de recursos essenciais:
# 'stopwords': lista de palavras irrelevantes (artigos, preposições).
# 'punkt_tab': parâmetros dos tokenizadores modernos de frases e palavras.
# 'machado': o corpus contendo as obras completas de Machado de Assis.
nltk.download('stopwords')
nltk.download("punkt_tab", raise_on_error=True)
nltk.download('machado')
```

```python
from nltk.corpus import machado
# Lista todos os identificadores de arquivos (obras) disponíveis no corpus Machado.
# Isso ajuda a localizar o caminho correto para carregar um livro específico.
machado.fileids()
```

```python
# Carrega o texto bruto (raw) do livro 'Memórias Postumas de Brás Cubas' (marm05.txt).
texto = machado.raw('romance/marm05.txt')

# Seleciona um trecho específico (slicing) e remove quebras de linha para facilitar a leitura.
# Aqui pegamos o início do Capítulo I.
texto = texto[10082:10954].replace('\n',' ')
display(texto)
```

```python
from nltk.tokenize import sent_tokenize
# Tokenização de Sentenças: divide o texto em frases individuais baseando-se na pontuação.
# O parâmetro language='portuguese' garante que as regras de pontuação do PT-BR sejam usadas.
sentencas = sent_tokenize(texto, language='portuguese')
for i, sent in enumerate(sentencas):
     print(i, '-', sent)
```

```python
from nltk.tokenize import word_tokenize
# Tokenização de Palavras: quebra o texto em unidades menores (palavras e símbolos de pontuação).
palavras = word_tokenize(texto, language='portuguese')
print(palavras)
```

```python
from nltk.tokenize import sent_tokenize, word_tokenize
# Processo combinado: primeiro dividimos em frases e, para cada frase, extraímos as palavras e símbolos.
sentencas = sent_tokenize(texto, language='portuguese')
for i, sent in enumerate(sentencas):
     print(i, '-', word_tokenize(sent, language='portuguese'))
```

```python
import string
# Obtém uma lista de caracteres de pontuação padrão do Python.
# Em NLP, remover pontuação ajuda a reduzir o ruído nos dados para análise estatística.
punctuations = list(string.punctuation)
print(punctuations)
```

```python
from nltk.tokenize import sent_tokenize, word_tokenize
# Limpeza de dados: removemos a pontuação de cada frase tokenizada.
sentencas = sent_tokenize(texto, language='portuguese')
for i, sent in enumerate(sentencas):
    # Compreensão de lista para filtrar apenas o que não for pontuação.
    sent_sem_pontuacao = [palavra for palavra in word_tokenize(sent, language='portuguese') if palavra not in punctuations]
    print(i, '-', sent_sem_pontuacao)
```

```python
# Carrega as Stopwords: palavras muito comuns (como 'o', 'a', 'de') que geralmente não agregam
# valor semântico para modelos de Machine Learning ou busca de informação.
stopwords = nltk.corpus.stopwords.words('portuguese')
print(stopwords)
```

```python
from nltk.tokenize import sent_tokenize, word_tokenize
# Pipeline de pré-processamento completo:
# 1. Tokenização -> 2. Remoção de Pontuação -> 3. Remoção de Stopwords.
sentencas = sent_tokenize(texto, language='portuguese')
for i, sent in enumerate(sentencas):
    sent_sem_pontuacao = [w for w in word_tokenize(sent, language='portuguese') if w not in punctuations]
    sent_sem_stopwords = [w for w in sent_sem_pontuacao if w not in stopwords]
    print(i, '-', sent_sem_stopwords)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
# Bag of Words (BoW): Transforma o texto em vetores numéricos.
# Cada posição no vetor representa a contagem de uma palavra do vocabulário na frase.
# ngram_range=(1,1) significa que estamos olhando para palavras únicas (unigramas).
cv = CountVectorizer(min_df=1, max_df=1.0, binary=False, ngram_range=(1,1))
bow = cv.fit_transform(sentencas)

print('BoW Shape:', bow.shape) # Exibe (número de frases, tamanho do vocabulário)
print('BoW Vocabulário (', len(cv.vocabulary_), 'palavras ):', cv.vocabulary_)
print('Primeira sentença vetorizada:', bow[0].toarray())
```
