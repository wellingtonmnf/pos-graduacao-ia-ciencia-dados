```python
# Demonstração acadêmica: ajuste e avaliação nos mesmos três pontos.
# As métricas abaixo descrevem o treino; não estimam generalização.
# Saídas históricas preservadas, sem recálculo nesta manutenção.
from sklearn import linear_model
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# gerando os dados
X = [[0, 0], [1, 1], [2, 2]]
Y = [0, 1, 2]
```

```python
# Criação do regressor linear ajustado por gradiente descendente estocástico (SGD).
clf = linear_model.SGDRegressor(alpha=0.1, max_iter=20)
```

```python
# treina o modelo
clf.fit (X, Y)
```

```python
# exibe os coeficientes do modelo
clf.coef_
```

```python
# gera previsões sobre os dados de treino
Y_pred = clf.predict(X)
```

```python
# avalia o erro das previsões
print('MSE no treino: %.2f' % mean_squared_error(Y, Y_pred))
print('R2 no treino: %.2f' % r2_score(Y, Y_pred))
```
