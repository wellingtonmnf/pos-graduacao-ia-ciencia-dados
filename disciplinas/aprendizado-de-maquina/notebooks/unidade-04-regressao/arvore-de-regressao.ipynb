```python
# Demonstração acadêmica: ajuste e avaliação nos mesmos três pontos.
# As métricas abaixo descrevem o treino; não estimam generalização.
# Saídas históricas preservadas, sem recálculo nesta manutenção.
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# gerando os dados
X = [[0, 0], [1, 1], [2, 2]]
Y = [0, 1, 2]
```

```python
# criação do modelo com o Arvore de Regressão
clfT = DecisionTreeRegressor(criterion="squared_error")
clfF = RandomForestRegressor(criterion="squared_error", n_estimators=10)
```

```python
# treina o modelo
clfT.fit (X, Y)
clfF.fit (X, Y)
```

```python
# gera previsões sobre os dados de treino
Y_pred_T = clfT.predict(X)
Y_pred_F = clfF.predict(X)
```

```python
# avalia o erro das previsões
print('MSE Árvore no treino: %.2f' % mean_squared_error(Y, Y_pred_T))
print('MSE Floresta no treino: %.2f' % mean_squared_error(Y, Y_pred_F))
print('R2 Árvore no treino: %.2f' % r2_score(Y, Y_pred_T))
print('R2 Floresta no treino: %.2f' % r2_score(Y, Y_pred_F))
```
