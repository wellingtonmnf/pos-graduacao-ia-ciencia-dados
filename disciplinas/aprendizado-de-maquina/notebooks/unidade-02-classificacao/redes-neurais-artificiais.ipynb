```python
# Revisão documental — 20/10/2026.
# Saídas históricas preservadas. C6 ainda mostra “Precisão” e C8 conserva a interpretação antiga dos
# pesos; a fonte atual descreve acurácia no treino e limita as conclusões sobre importância e
# causalidade. Execução integral desta revisão no Colab pendente.

# 1. IMPORTAR AS BIBLIOTECAS NECESSÁRIAS
import numpy as np
import pandas as pd
from sklearn.linear_model import Perceptron
```

```python
# 2. CRIAR OS DADOS DA TABELA MANUALMENTE
# Criando um dicionário com os dados dos alunos
dados = {
    'Aluno': ['João', 'Maria', 'José'],
    'Idade': [20, 25, 30],
    'Tempo_de_Estudo': [4, 2, 1],
    'Faltas': [4, 2, 6],
    'Prova_1': [10, 9, 1]
}

# Transformando em um DataFrame do pandas (uma tabela)
df = pd.DataFrame(dados)

print("=" * 50)
print("DADOS DOS ALUNOS")
print("=" * 50)
print(df)
print("\n")
```

```python
# 3. PREPARAR OS DADOS PARA O TREINAMENTO
# X = features (características que vamos usar para prever)
# Vamos usar: Idade, Tempo de Estudo e Faltas
X = df[['Idade', 'Tempo_de_Estudo', 'Faltas']].values

# y = target (o que queremos prever)
# Precisamos criar uma classificação binária:
# 1 = desempenho superior a 70% (mais de 8.4 pontos)
# 0 = desempenho inferior ou igual a 70%
nota_maxima = 12
limiar = nota_maxima * 0.70  # 70% de 12 = 8.4 pontos

y = (df['Prova_1'] > limiar).astype(int).values

print("=" * 50)
print("DADOS PREPARADOS PARA TREINAMENTO")
print("=" * 50)
print(f"Limiar para aprovação: {limiar} pontos (70% de {nota_maxima})")
print(f"\nX (Features - Entrada):\n{X}")
print(f"\ny (Target - Saída Desejada):\n{y}")
print(f"\nInterpretação do y:")
for i, aluno in enumerate(df['Aluno']):
    status = "APROVADO" if y[i] == 1 else "REPROVADO"
    print(f"  {aluno}: Nota {df['Prova_1'][i]} = {status} (70%+)")
print("\n")
```

```python
# 4. CRIAR E TREINAR O PERCEPTRON
# Transpor os dados (o np.rot90 fazia isso, mas vamos usar .T que é mais claro)
# Na verdade, neste caso não precisamos rotacionar

net = Perceptron(max_iter=100, eta0=0.1, random_state=42)
net.fit(X, y)

print("=" * 50)
print("TREINAMENTO CONCLUÍDO")
print("=" * 50)
```

```python
# 5. FAZER PREVISÕES E AVALIAR O MODELO
previsoes = net.predict(X)
acuracia = net.score(X, y) * 100

print(f"\nSaída Esperada (Real):     {y}")
print(f"Saída Prevista (Modelo):   {previsoes}")
print(f"\nAcurácia no treino (3 exemplos): {acuracia:.2f}%")
print(f"\nPesos do modelo: {net.coef_}")
print(f"Bias (intercepto): {net.intercept_}")

# score retorna acurácia. As saídas históricas abaixo não foram recalculadas.

# X e y são os mesmos três exemplos usados no ajuste.
# Esta acurácia descreve o ajuste e não estima generalização em alunos novos.
```

```python
# 6. ANÁLISE DETALHADA DOS RESULTADOS
print("\n" + "=" * 50)
print("ANÁLISE DETALHADA POR ALUNO")
print("=" * 50)
for i, aluno in enumerate(df['Aluno']):
    real = "SIM" if y[i] == 1 else "NÃO"
    previsto = "SIM" if previsoes[i] == 1 else "NÃO"
    acertou = "✓" if y[i] == previsoes[i] else "✗"

    print(f"\n{aluno}:")
    print(f"  Faltas: {df['Faltas'][i]} | Tempo Estudo: {df['Tempo_de_Estudo'][i]}h | Idade: {df['Idade'][i]}")
    print(f"  Nota Real: {df['Prova_1'][i]}/12")
    print(f"  Desempenho > 70%? Real: {real} | Previsto: {previsto} {acertou}")
```

```python
# 7. INTERPRETAÇÃO DOS PESOS
print("\n" + "=" * 50)
print("INTERPRETAÇÃO DOS PESOS")
print("=" * 50)
print(f"Peso da Idade: {net.coef_[0][0]:.4f}")
print(f"Peso do Tempo de Estudo: {net.coef_[0][1]:.4f}")
print(f"Peso das Faltas: {net.coef_[0][2]:.4f}")
print("\nOs coeficientes atuam na função de decisão linear, junto do intercepto.")
print("As entradas estão em suas escalas originais, sem padronização; magnitudes absolutas não são uma comparação direta de importância.")
print("Mantidas as outras entradas, um coeficiente negativo reduz o score quando aquela entrada aumenta.")
print("Neste classificador binário, scores positivos favorecem a classe 1 (desempenho acima de 70%); o modelo não prevê uma nota numérica.")
print("Essa relação no modelo não demonstra efeito causal sobre o desempenho.")
# Explicação corrigida; saídas históricas preservadas, sem nova execução do treino.
```
